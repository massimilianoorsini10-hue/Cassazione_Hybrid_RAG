```python
import json
from tqdm import tqdm

from transformers import AutoTokenizer, AutoModelForTokenClassification, pipeline
import torch
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
tokenizer_chunk = AutoTokenizer.from_pretrained("dlicari/Italian-Legal-BERT")

text_splitter = RecursiveCharacterTextSplitter.from_huggingface_tokenizer(tokenizer_chunk, chunk_size=400, chunk_overlap=50)

with open("database_sentenze.json", 'r', encoding = 'utf-8') as j:
    database = json.load(j)


nuovo_database_sentenze = []

for sentenza in tqdm(database):
    if sentenza is None:
        continue
    else:
        nuovo_testo = sentenza['testo_completo'].replace('___','')
        nuovo_json = {'ID': sentenza['ID'],
                      'metadati': sentenza['metadati'],
                      'testo' : nuovo_testo,
                      'chunks' : {}}
        chunks = text_splitter.split_text(nuovo_testo)
        
        nuovo_database_sentenze.append(nuovo_json)

nuovo_percorso = "/Users/maxiorsini/Cassazione_Hybrid_RAG/nuovo_database_sentenze.json"

with open(nuovo_percorso, 'w', encoding='utf-8') as nj:
    json.dump(nuovo_database_sentenze, nj, ensure_ascii=False, indent=4)
```
