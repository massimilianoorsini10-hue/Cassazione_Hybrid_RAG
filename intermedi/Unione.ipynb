```python
import json
```

```python
path_database = 'database_sentenze_ordinato.json'
path_riferimenti = 'riferimenti_ordinati.json'

with open(path_database, 'r', encoding='utf-8') as d:
    database=json.load(d)

with open(path_riferimenti, 'r', encoding='utf-8') as t:
    riferimenti=json.load(t)
```

```python
for i, sentenza in enumerate(riferimenti):
    if database[i]['ID'] == sentenza['sentenza_id']:
        for chunk in sentenza['chunks']:
            for c in database[i]['chunks']:
                if int(c.split('_')[1]) == chunk['chunk_id']:
                    database[i]['chunks'][c]['riferimenti'] = chunk['riferimenti']


```

```python
with open('database_chunk_riferimenti.json', 'w', encoding='utf-8') as dcr:
    json.dump(database, dcr, ensure_ascii=False, indent=4)
```
